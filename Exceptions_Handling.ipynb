## Exception Handling

```python
# In this file we learn how to handle the exception in differnet way and how to store information in txt and log file.
```

```python
try:
    c=10/0
except ZeroDivisionError as e:
    print("This is ZeroDivisionError")
```

```python
import logging
logging.basicConfig(filename='exception.log',level=logging.DEBUG,format='%(asctime)s - %(message)s - %(levelname)s')
```

```python
try:
    c=10/0
except ZeroDivisionError as e:
    logging.error("In this code user if user try to divide any number with zero it will genrate error{}".format(e))
```

```python
try:
    with open('exception.txt','r') as f:
        t=f.read()
except FileNotFoundError as e:
    logging.info("The error is genrated because the file is not{}".format(e))
```

```python
try:
    with open('exception.txt','w') as f:
        f.write("The file is created for store the information on any code")
except FileNotFoundError as e:
    logging.info("The error is genrated because the file is not{}".format(e))
finally:
    f.close() # in any condition the open file is close at the end so we write this in finally block
```
