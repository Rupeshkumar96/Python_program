```python
class BankAccount():
    def __init__(self,balance):
        self.__balance=balance
    def display(self):
        print(f'Total Amount in your account is:{self.__balance}')
    def deposite(self,amount):
        self.__balance+=amount
    def withdraw(amount):
        if self.__balance>1000:
            self.__balance-=self.__balance-amount
        else:
            print("You have low balance, you can't widthraw the amount")
```

```python
m=BankAccount(50000)
m.deposite(60)
m.display()
```

## INHERITANCE

```python
# single inheritance
class test():
    def name(self):
        print("Enter your name")
    def details(self):
        print("Enter your details")
class test1(test):
    pass
```

```python
test2=test1
test2.name()
```

```python
# Multiple Inheritance
class test():
    def name(self):
        print("Enter your name")
    def details(self):
        print("Enter your details")
class test1():
     def address(self):
         print("Rampur")
class test2(test,test1):
    pass
```

```python
t=test2()
t.address()
t.name()
```

```python
# Multilevel Inheritance
class test():
    def name(self):
        print("Enter your name")
    def details(self):
        print("Enter your details")
class test1(test):
     def address(self):
         print("Rampur")
class test2(test1):
    pass
```

```python
t=test2()
t.address()
t.name()
```

## Abstract Class:- Blueprint of the code / Program

```python
import abc
class test(abc.ABC):
    @abc.abstractmethod
    def name(self):
        pass
    @abc.abstractmethod
    def address(self):
        pass
    @abc.abstractmethod
    def result(self):
        pass
```

```python
class inherit(test):
    def name(self):
        print("Rupesh")
    def address(self):
        print("Rampur")
    def result(self):
        print("passed")
```

```python
h=inherit()
h.address()
h.name()
```

```python
import abc
class test(abc.ABC):
    @classmethod
    @abc.abstractmethod
    def name(cls):
        pass
    @abc.abstractmethod
    def add(self):
        pass
class test1(test):
    @classmethod
    def name(cls):
        print("Rupesh")
    def add(self):
        print("Rampur")
```

```python
test1.name()
t=test1()
t.name()
t.add()
```

```python
from abc import ABC, abstractmethod

class A(ABC):

    @classmethod
    @abstractmethod
    def show(cls):
        pass


class B(A):

    @classmethod
    def show(cls):
        print("This is class method in B")


# ✔️ Object create
obj = B()

# ✔️ Call using class
B.show()

# ✔️ Call using object
obj.show()
```

## Property Decorator Method

```python
class movie:
    def __init__(self,name,tprice):
        self.__name=name
        self.__tprice=tprice
    @property
    def Display(self):
        return self.__tprice
    @Display.setter
    def change_price(self,price):
        if self.__tprice<500:
            pass
        else:
            self.__tprice=price
    
```

```python
m=movie("Spiderman",500)
m.Display
```
