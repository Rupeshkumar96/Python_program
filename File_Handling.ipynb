## File Handling

```python
file=open("test.txt",'w') # File open and write
file.write("Hello kaise ho")
file.close()      # File close after use
```

```python
with open('test.txt','a') as f:
    f.write("This is updated content")    # If we use the with then we not to  close manually 
```

```python
# In this we read the file
with open('test.txt','r') as f:
    t=f.read()
    print(t)
```

```python
# USE OF READ THIS IS READ ALL THE CONTENT
file=open('test.txt','r')
t=file.read()
print(t)
file.close()
```

```python
# READLINE IT IS USE THE READ THE CONTENT LINE BY LINE
file=open('test.txt','r')
t=file.readline()
print(t)
file.close()
```

```python
# READLINES IT IS USE TO READ THE ALL THE LINE
file=open('test.txt','r')
t=file.readlines()
print(t)
file.close()
```

```python
# WRITELINES USE FOR WRITE THE MULTIPLE LINES
with open('test1.txt','w') as f:
    f.writelines("""A circular queue is a special type of queue data structure in which the last position is connected back to the first position 
    forming a circle. Unlike a simple linear queue, it efficiently utilizes memory by reusing empty spaces created when elements are dequeued.""")
```

## CODE WITH OPERATING SYSTEM FUNCTION

```python
import os
```

```python
# Check the file is exist  or not
if os.path.exists("test1.txt"):
    print("File is exists")
else:
    print("File is not exists")
```

```python
# If we want to remove the exists file then use
if os.path.exists("test.txt"):
    print("file is exits")
    os.remove('test.txt')
else:
    print("File is not exists")
```

## Work with json file

```python
import json
```

```python
data={'name':'rupesh','roll':55,'class':'mca'}
```

```python
# json.dumps() --> it is used to convert dictionary object to json file.
json_string = json.dumps(data)
print(json_string) 
print(type(json_string)) 
```

```python
# json.dump --> it is used to write the json file
data={'name':'rupesh','roll':55,'class':'mca'}
with open('test2.json','w') as f:
    json.dump(data,f)
```

```python
# json.load --> it is used to read the content of json file
with open('test2.json','r') as f:
    t=json.load(f)
print(t)
```

```python
# json.loads --> it is used to convert json file into python object
json_data = '{"name": "Alice", "age": 25, "city": "New York"}'
python_obj = json.loads(json_data)
print(python_obj) 
print(type(python_obj))
```
